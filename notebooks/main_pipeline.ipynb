#### 3. Load bản đồ khu vực trường Đại học Bách Khoa

```python
from modules.graph.map_loader import load_map

LAT  = 10.7729
LON  = 106.6578
DIST = 2000  # bán kính 2km

graph_data = load_map(LAT, LON, DIST)

G     = graph_data["G"]
nodes = graph_data["nodes"]
edges = graph_data["edges"]

print(f"✅ Load bản đồ thành công!")
print(f"   Số nodes: {len(nodes)}")
print(f"   Số edges: {len(edges)}")
```

#### 4. Khởi tạo MapGraph và DeliveryVehicle

```python
import osmnx as ox
from modules.graph.core_system import MapGraph, DeliveryVehicle

# Khởi tạo MapGraph từ dữ liệu bản đồ
map_graph = MapGraph(graph_data)
print(f"✅ Khởi tạo MapGraph thành công!")
print(f"   Số node trong đồ thị: {len(map_graph.G.nodes)}")
print(f"   Số cạnh trong đồ thị: {len(map_graph.G.edges)}")

# ─────────────────────────────────────────────
# Ghim điểm xuất phát và 3 điểm giao hàng
# bằng tọa độ GPS thực tế (kinh độ, vĩ độ)
# → dùng nearest_nodes để ánh xạ sang node OSM
# ─────────────────────────────────────────────
LOCATIONS = {
    "origin" : (106.6578, 10.7729),  # ĐH Bách Khoa
    "dest_A" : (106.6600, 10.7750),  # Ngã tư Đinh Tiên Hoàng
    "dest_B" : (106.6555, 10.7800),  # Khu vực Võ Thị Sáu
    "dest_C" : (106.6520, 10.7690),  # Khu vực Lý Thường Kiệt
    "dest_D" : (106.6650, 10.7710),  # Khu vực Nguyễn Thị Minh Khai
}

lons = [coord[0] for coord in LOCATIONS.values()]
lats = [coord[1] for coord in LOCATIONS.values()]
nearest = ox.nearest_nodes(map_graph.G, X=lons, Y=lats)

origin_node, dest_A, dest_B, dest_C, dest_D = nearest
destinations = [dest_A, dest_B, dest_C, dest_D]

# Khởi tạo xe giao hàng
# Xuất phát lúc 8h sáng (480 phút), tải trọng 2.5 tấn
vehicle = DeliveryVehicle(
    start_node=origin_node,
    start_time=480.0,
    capacity=2.5
)

# Thêm các điểm giao hàng với khung giờ
vehicle.add_delivery_point(dest_A, time_window=(480.0, 540.0))  # 60 phút
vehicle.add_delivery_point(dest_B, time_window=(480.0, 530.0))  # 50 phút
vehicle.add_delivery_point(dest_C, time_window=(490.0, 550.0))  # 60 phút
vehicle.add_delivery_point(dest_D, time_window=(495.0, 560.0))  # 65 phút

print(f"\n✅ Khởi tạo DeliveryVehicle thành công!")
print(f"   Điểm xuất phát : {origin_node} (ĐH Bách Khoa)")
print(f"   dest_A         : {dest_A} (Đinh Tiên Hoàng)")
print(f"   dest_B         : {dest_B} (Võ Thị Sáu)")
print(f"   dest_C         : {dest_C} (Lý Thường Kiệt)")
print(f"   dest_D         : {dest_D} (Nguyễn Thị Minh Khai)")
print(f"   Giờ xuất phát  : {vehicle.current_time} phút (= 8:00 sáng)")
print(f"   Tải trọng xe   : {vehicle.capacity} tấn")
```

#### 5. Lọc đồ thị theo luật logic giao thông

```python
from modules.bayes_logic.logic_filter import filter_graph

# Lọc các cạnh vi phạm luật giao thông
# vehicle.capacity = 2.5 tấn
cleaned_graph = filter_graph(graph_data, vehicle_weight=vehicle.capacity)

# Cập nhật đồ thị đã lọc vào MapGraph
map_graph.apply_logic_filter(cleaned_graph)

# In thống kê kết quả lọc
stats = cleaned_graph.graph["logic_filter_stats"]
print(f"✅ Lọc đồ thị thành công!")
print(f"   Tải trọng xe             : {stats['vehicle_weight_ton']} tấn")
print(f"   Cạnh bị xóa (tổng)       : {stats['removed_total']}")
print(f"   - Độ dài không hợp lệ    : {stats['removed_invalid_length']}")
print(f"   - Bị cấm truy cập        : {stats['removed_restricted_access']}")
print(f"   - Vượt giới hạn tải trọng: {stats['removed_weight_limit']}")
print(f"   - Đường đóng/xây dựng   : {stats['removed_closed_or_construction']}")
print(f"   Số cạnh còn lại          : {stats['remaining_edges']}")
```

#### 6. Tính xác suất kẹt xe bằng mạng Bayes

```python
from modules.bayes_logic.bayes_model import compute_congestion

# Thiết lập điều kiện môi trường
WEATHER     = "rain"   # Thời tiết: "rain" hoặc "clear"
TIME_OF_DAY = "peak"   # Khung giờ: "peak" hoặc "normal"

# Tính xác suất kẹt xe cho từng cạnh
congestion_df = compute_congestion(
    edges=edges,
    weather=WEATHER,
    time_of_day=TIME_OF_DAY
)

# Thêm thông tin môi trường vào congestion_df để TV3 dùng
congestion_df["weather"]     = WEATHER
congestion_df["time_of_day"] = TIME_OF_DAY

print(f"✅ Tính xác suất kẹt xe thành công!")
print(f"   Điều kiện : thời tiết={WEATHER}, khung giờ={TIME_OF_DAY}")
print(f"   Số cạnh   : {len(congestion_df)}")
print(f"\n📊 Thống kê xác suất kẹt xe:")
print(f"   Trung bình : {congestion_df['p_congestion'].mean():.4f}")
print(f"   Thấp nhất  : {congestion_df['p_congestion'].min():.4f}")
print(f"   Cao nhất   : {congestion_df['p_congestion'].max():.4f}")
print(f"\n5 cạnh có xác suất kẹt xe cao nhất:")
print(congestion_df.nlargest(5, 'p_congestion')[['u', 'v', 'p_congestion']].to_string(index=False))
```

#### 7. Dự đoán thời gian di chuyển bằng ML

```python
from modules.ml.travel_time_predictor import predict_travel_time

# Dự đoán thời gian di chuyển cho từng cạnh
# Kết hợp dữ liệu edges (TV1) + congestion_df (TV4) → weighted_edges (TV3)
weighted_edges = predict_travel_time(
    edges=edges,
    congestion_df=congestion_df
)

print(f"✅ Dự đoán thời gian di chuyển thành công!")
print(f"   Số cạnh dự đoán: {len(weighted_edges)}")
print(f"\n📊 Thống kê thời gian di chuyển (phút):")
print(f"   Trung bình : {weighted_edges['travel_time_min'].mean():.4f}")
print(f"   Thấp nhất  : {weighted_edges['travel_time_min'].min():.4f}")
print(f"   Cao nhất   : {weighted_edges['travel_time_min'].max():.4f}")
print(f"\n5 cạnh có thời gian di chuyển cao nhất:")
print(weighted_edges.nlargest(5, 'travel_time_min')[['u', 'v', 'travel_time_min']].to_string(index=False))
```

#### 8. Tìm lộ trình tối ưu bằng A* và CSP

```python
from modules.search.astar import run_astar

# Chạy thuật toán A*
result = run_astar(
    cleaned_graph=map_graph.G,
    weighted_edges=weighted_edges,
    origin=origin_node,
    destinations=destinations,
    time_windows=vehicle.deliveries,
    start_time=vehicle.current_time
)

print(f"✅ Tìm lộ trình thành công!")
print(f"\n📍 Kết quả lộ trình:")
print(f"   Tổng thời gian   : {result['total_time_min']:.4f} phút")
print(f"   Số node lộ trình : {len(result['route'])} nodes")
```

#### 9. Tích hợp hệ thống & Mô phỏng Re-planning

```python
from modules.integration.coordinator import DeliveryCoordinator

# ─────────────────────────────────────────────
# Khởi tạo DeliveryCoordinator
# Constructor chỉ cần map_graph — BayesCongestionModel
# được khởi tạo nội bộ bên trong coordinator
# ─────────────────────────────────────────────
coordinator = DeliveryCoordinator(map_graph=map_graph)

print("✅ Khởi tạo DeliveryCoordinator thành công!")
print(f"   Thời tiết mặc định : {coordinator.current_weather}")
print(f"   Khung giờ mặc định : {coordinator.current_time_of_day}")
```

```python
import networkx as nx

def get_weight(u, v, d):
    if not d: return 9999.0
    return min(e.get('travel_time_min', 9999.0) for e in d.values())

# Inject tắc đường trên chặng dest_B → dest_C
# (chặng đầu tiên mà re-planning sẽ xét)
try:
    bottleneck_path = nx.shortest_path(
        cleaned_graph, dest_B, dest_C, weight=get_weight
    )
    bottleneck_edges = [
        (bottleneck_path[i], bottleneck_path[i+1])
        for i in range(min(3, len(bottleneck_path)-1))
    ]
    coordinator.set_local_congestion(bottleneck_edges, p_congestion=0.95)
    print(f"⚠️ Tắc đường cục bộ inject vào {len(bottleneck_edges)} cạnh trên chặng dest_B→dest_C")
    print(f"   bottleneck_edges: {bottleneck_edges}")
except nx.NetworkXNoPath:
    print("⚠️ Không tìm được đường để inject tắc đường cục bộ")

# ─────────────────────────────────────────────
# Lập kế hoạch lộ trình ban đầu
# Coordinator tự chạy TV4 (Bayes) → TV3 (ML) → TV2 (A*)
# ─────────────────────────────────────────────
initial_plan = coordinator.plan_initial_route(
    vehicle=vehicle,
    weather=WEATHER,
    time_of_day=TIME_OF_DAY,
)

if initial_plan and initial_plan.get("route"):
    print(f"\n✅ Kế hoạch ban đầu:")
    print(f"   Tổng thời gian   : {initial_plan['total_time_min']:.4f} phút")
    print(f"   Số node lộ trình : {len(initial_plan['route'])} nodes")
else:
    print("❌ Không tìm được lộ trình ban đầu!")
```

```python
# ─────────────────────────────────────────────
# Mô phỏng thay đổi môi trường → Re-planning
# ─────────────────────────────────────────────
NEW_WEATHER     = "clear"
NEW_TIME_OF_DAY = "normal"

coordinator.clear_local_congestion()
print("✅ Tắc đường cục bộ đã được giải tỏa")

replan_result = coordinator.trigger_replanning(
    vehicle=vehicle,
    new_weather=NEW_WEATHER,
    new_time_of_day=NEW_TIME_OF_DAY,
)

if replan_result and replan_result.get("route"):
    delivered_node = initial_plan["visited_order"][0]
    delivered_time = initial_plan["arrival_times"][delivered_node]
    delivered_idx = initial_plan["route"].index(delivered_node)
    delivered_idx = initial_plan["route"].index(delivered_node)
    remaining_initial_nodes = len(initial_plan["route"]) - delivered_idx
    remaining_initial_nodes = len(initial_plan["route"]) - delivered_idx
    remaining_initial_time = initial_plan["total_time_min"] - (delivered_time - 480.0)

    print(f"\n📊 So sánh lộ trình (chỉ xét {len(initial_plan['visited_order']) - 1} điểm chưa giao):")
    print(f"   Điểm đã giao thành công : {delivered_node} (lúc {delivered_time:.2f} phút)")
    print(f"   Xe đang ở               : Node {vehicle.current_location} | {vehicle.current_time:.2f} phút")
    print()
    print(f"   Số node lộ trình (kế hoạch cũ) : {remaining_initial_nodes} nodes")
    print(f"   Số node lộ trình (re-planning)  : {len(replan_result['route'])} nodes")
    print()
    print(f"   Thời gian còn lại (kế hoạch cũ): {remaining_initial_time:.4f} phút")
    print(f"   Thời gian re-planning           : {replan_result['total_time_min']:.4f} phút")
    delta = replan_result["total_time_min"] - remaining_initial_time
    sign  = "+" if delta >= 0 else ""
    print(f"   Thay đổi                        : {sign}{delta:.4f} phút")

elif replan_result is None:
    print("ℹ️ Xe đã hoàn thành tất cả điểm giao — không cần re-plan.")
else:
    print("❌ Re-planning thất bại — cần điều phối thủ công!")
```

#### 10. Visualization lộ trình tối ưu lên bản đồ

```python
import folium

# Tạo bản đồ nền
m = folium.Map(location=[LAT, LON], zoom_start=15)

# Vẽ toàn bộ mạng lưới đường bộ (màu xám nhạt)
for _, row in edges.iterrows():
    u_data = nodes[nodes["osmid"] == row["u"]]
    v_data = nodes[nodes["osmid"] == row["v"]]
    if not u_data.empty and not v_data.empty:
        u_coord = (u_data.iloc[0]["y"], u_data.iloc[0]["x"])
        v_coord = (v_data.iloc[0]["y"], v_data.iloc[0]["x"])
        folium.PolyLine(
            [u_coord, v_coord],
            color="gray", weight=1, opacity=0.3
        ).add_to(m)

# Vẽ lộ trình ban đầu A* (màu đỏ)
route = result["route"]
for i in range(len(route) - 1):
    u_data = nodes[nodes["osmid"] == route[i]]
    v_data = nodes[nodes["osmid"] == route[i + 1]]
    if not u_data.empty and not v_data.empty:
        u_coord = (u_data.iloc[0]["y"], u_data.iloc[0]["x"])
        v_coord = (v_data.iloc[0]["y"], v_data.iloc[0]["x"])
        folium.PolyLine(
            [u_coord, v_coord],
            color="red", weight=4, opacity=0.8,
            tooltip="Lộ trình ban đầu (mưa, peak)"
        ).add_to(m)

# Vẽ lộ trình re-planning (màu xanh dương)
if replan_result and replan_result.get("route"):
    replan_route = replan_result["route"]
    for i in range(len(replan_route) - 1):
        u_data = nodes[nodes["osmid"] == replan_route[i]]
        v_data = nodes[nodes["osmid"] == replan_route[i + 1]]
        if not u_data.empty and not v_data.empty:
            u_coord = (u_data.iloc[0]["y"], u_data.iloc[0]["x"])
            v_coord = (v_data.iloc[0]["y"], v_data.iloc[0]["x"])
            folium.PolyLine(
                [u_coord, v_coord],
                color="blue", weight=4, opacity=0.8,
                tooltip="Lộ trình re-planning (quang đãng, normal)"
            ).add_to(m)

# Đánh dấu điểm xuất phát (màu xanh lá)
origin_data = nodes[nodes["osmid"] == origin_node].iloc[0]
folium.Marker(
    location=[origin_data["y"], origin_data["x"]],
    popup=f"Xuất phát: {origin_node}",
    icon=folium.Icon(color="green", icon="play", prefix="fa")
).add_to(m)

# Đánh dấu các điểm giao hàng (màu đỏ)
for i, dest in enumerate(result["visited_order"]):
    dest_data = nodes[nodes["osmid"] == dest].iloc[0]
    folium.Marker(
        location=[dest_data["y"], dest_data["x"]],
        popup=f"Điểm giao {i+1}: {dest}",
        icon=folium.Icon(color="red", icon="star", prefix="fa")
    ).add_to(m)

print(f"✅ Visualization thành công!")
print(f"   Đường xám  : toàn bộ mạng lưới đường bộ")
print(f"   Đường đỏ   : lộ trình ban đầu (mưa, peak)")
print(f"   Đường xanh : lộ trình re-planning (quang đãng, normal)")
print(f"   Ghim xanh  : điểm xuất phát")
print(f"   Ghim đỏ    : các điểm giao hàng")
m
```
